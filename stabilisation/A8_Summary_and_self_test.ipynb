## 8.2 What the calculations in these notebooks do not include

* **Ion pairs.** Part of the calcium is bound to bicarbonate or carbonate as dissolved pairs (CaHCO₃⁺, and the uncharged CaCO₃⁰), which lowers the free Ca²⁺ and CO₃²⁻ slightly. **PHREEQC** – a free geochemical calculation program from the US Geological Survey, widely used for water chemistry – includes these and many more species. In soft water the difference to our calculation is a few hundredths in SI.
* **Other buffers** (NOM, silicate, phosphate) – humic water has some extra buffer from organic acids.
* **Kinetics:** a marble filter approaches equilibrium but does not reach it; designs use contact time and experience.
* **Copper and lead release**, which depend on more than saturation (NOM, stagnation, flow).

State these limits when you use the method in your semester project.

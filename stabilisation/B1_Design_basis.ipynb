## 1.1 Every number needs a source

A design report is checked by others: the client, the authorities, the contractor. Each of them must be able to see **where a value comes from**. In this example every value is one of three kinds:

1. **From a guideline or the lecture** – e.g. *NV 212/2015 Table 12.3* or *lecture M08, slide 'Target water quality'*. Give the table or section, not just the report.
2. **Calculated** – from other values with an equation that is shown.
3. **An assumption** – a value we chose because no better information is available. It is marked as <span style='background:rgba(255,167,38,0.25);padding:0 4px;border-radius:3px'>assumption</span> and listed in the **assumption register** (1.5). Prices are marked as <span style='background:rgba(229,57,53,0.22);padding:0 4px;border-radius:3px'>ESTIMATE</span>.

The same rule applies in your semester project: a design value without a verifiable source is not accepted – also not if an AI tool suggested it.

**Abbreviations used for sources:** *NV 212* = Norsk Vann rapport 212/2015, *Veiledning for dimensjonering av vannbehandlingsanlegg* (section § or table). *NV 274* = Norsk Vann rapport 274/2022, *Korrosjonsbeskyttelse – erfaring og ny kunnskap*. *M08* = lecture M08 Stabilisation and corrosion control (slide title).

## 1.2 Flows: which flow for what?

A water treatment plant does not see one flow, but a range. For the example we use (<span style='background:rgba(255,167,38,0.25);padding:0 4px;border-radius:3px'>assumption</span> – in your project the flows come from the water demand of the supply area):

| Flow | Value | Meaning |
| --- | --- | --- |
| Q<sub>avg</sub> | 50 L/s = 180 m³/h = 4 320 m³/d | average over the year |
| Q<sub>max</sub> | 80 L/s = 288 m³/h | highest hourly flow on the day of highest demand |
| Q<sub>min</sub> | 15 L/s = 54 m³/h | lowest flow (night, low season) |

**Design flow.** NV 212 (ch. 2, *Dimensjonerende vannmengde*) defines the design flow of a treatment plant as **the maximum hourly flow on the maximum day**: Q<sub>dim</sub> = Q<sub>max hour in max day</sub>. Only if an equalisation tank lies between the treatment step and the consumers may the *mean* hour of the maximum day be used. We assume that the marble filter sees the full Q<sub>max</sub>, so **Q<sub>dim</sub> = 80 L/s = 288 m³/h**.

Each design task needs a different flow – a frequent mistake is to use Q<sub>avg</sub> for everything:

| Design task | Flow | Why |
| --- | --- | --- |
| size of the marble filters (area, EBCT) | Q<sub>max</sub>, with one filter out of service | the contact time must be enough at the worst moment (B4) |
| maximum capacity of the CO₂ dosing | Q<sub>max</sub> × highest dose | the dosing must keep up at peak flow (B5) |
| minimum capacity of the CO₂ dosing | Q<sub>min</sub> × lowest dose | the dosing must still be accurate at night (B5) |
| chemical consumption, storage, costs | Q<sub>avg</sub> | what is used over weeks and years (B5) |

## 1.3 The raw water

The lake water was sampled several times; the design values (measured in the laboratory at 5 °C) are:

| Parameter | Design value | Assumed range over the year | Comment |
| --- | --- | --- | --- |
| pH | 6.6 | 6.4–6.8 | lowest at snowmelt |
| Alkalinity | 0.10 mmol/L | 0.05–0.15 mmol/L | very low – almost no buffer |
| Calcium | 2.5 mg/L | – | typical for Norwegian surface water: 1–5 mg/L (M08, slide 'Typical Norwegian drinking water') |
| Conductivity | 35 µS/cm | – | used for the ionic strength (chapter A5) |
| Temperature | – | 2–12 °C | 2 °C in winter, 12 °C in late summer |

The ranges and the temperature are <span style='background:rgba(255,167,38,0.25);padding:0 4px;border-radius:3px'>assumption</span>s for this example; they are typical for a Norwegian lake (M08, slide 'Target water quality': raw surface water pH 6–7, 1–5 mg Ca/L, alkalinity often below 0.2 mmol/L). From pH and alkalinity the total inorganic carbon follows (chapter A4): **C<sub>T</sub> = 0.18 mmol/L**.

A real design basis would be built on a year of monitoring data, and use e.g. the 10 % and 90 % values rather than the mean.

```python
dv.basis_panel()
```

**What the panel shows.** The raw water lies far below and to the left of the target zone: it lacks alkalinity (vertical distance) and inorganic carbon (horizontal distance), and its calcium (2.5 mg/L) is far from 15–25 mg/L. The saturation index is around −3.8: the water dissolves CaCO₃ from cement and cannot build a protective layer. Note how the pH of the *same* water changes from winter to summer although nothing is added – the equilibrium constants change with temperature (chapter A3). This is why the design uses C<sub>T</sub> and alkalinity, and not the pH, as the variables that describe the water.

## 1.4 The target water quality

The targets come from **NV 274/2022** (lecture M08, slide 'Target water quality'):

| Parameter | Target | Why (M08) |
| --- | --- | --- |
| pH | 8.0–8.5 | above the saturation pH; low copper release |
| Alkalinity | 0.6–1.0 mmol/L (≈ 1.0) | buffer – a stable pH in the network |
| Calcium | 15–25 mg Ca/L (≈ 20) | CaCO₃ layer, SI ≈ 0 |

NV 212/2015 §12.2.1 still gives pH 8.0–9.0 (from the Mattilsynet guidance of 2004); NV 274/2022 narrowed it to 8.0–8.5, and we use the newer value. Because the network is mixed (cast iron, cement, copper), we aim for the **compromise for mixed networks** (M08, slide 'Targets per pipe material'): copper prefers a high pH and *little* inorganic carbon, iron and cement prefer *much* inorganic carbon and calcium. One water for all materials is always a compromise.

## 1.5 Assumption register

Every assumption is written down, with the reason and what would have to be checked for a real plant. The list grows in the following chapters; the complete register is in B8.

| # | Assumption | Value | Why this value | Check for a real plant |
| --- | --- | --- | --- | --- |
| A1 | flows | 50 / 80 / 15 L/s | example | water demand, consumption data |
| A2 | no equalisation before the marble filter | Q<sub>dim</sub> = Q<sub>max</sub> | conservative | plant layout |
| A3 | temperature | 2–12 °C | typical deep lake | temperature logger at the intake |
| A4 | raw-water range | pH 6.4–6.8, Alk 0.05–0.15 mmol/L | typical seasonal variation | one year of monitoring |
| A5 | no coagulation needed | – | low colour | colour and TOC data, barrier analysis |
| A6 | mixed network | cast iron, cement, plastic, copper | old network | network register |

## Key points

<div style='background:rgba(67,160,71,0.12);border-left:4px solid #43a047;padding:8px 14px'>

- A design basis contains flows, raw water quality over the year, targets – and a source or an assumption for each value.
- Q<sub>dim</sub> = Q<sub>max</sub> (max hour, max day; NV 212) sizes the units; Q<sub>avg</sub> gives consumption and costs; Q<sub>min</sub> the smallest dosing rate.
- Targets (NV 274/2022): pH 8.0–8.5, alkalinity 0.6–1.0 mmol/L, calcium 15–25 mg/L.
- Describe the water with alkalinity and C<sub>T</sub>; the pH changes with temperature.

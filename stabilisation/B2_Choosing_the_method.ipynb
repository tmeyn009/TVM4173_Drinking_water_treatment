## 2.1 The options

Chapter A7 described the methods in use in Norway (lecture M08, slides 'Three strategies in Norway' and 'Choosing a method'). For our water we compare them with five questions:

| Method | Adds Ca? | Adds alkalinity? | Fits our water? | Main drawback |
| --- | --- | --- | --- | --- |
| **CO₂ + marble filter** | yes | yes | yes – NV 212 §12.2.2.2: used where little other treatment is needed | large filters; marble must be refilled |
| lime + CO₂ | yes | yes | possible | lime handling (dust, lime saturator, turbidity); usual where coagulation needs pH control |
| NaOH only | no | yes | no – the network is not all plastic | no calcium, no CaCO₃ layer on iron and cement |
| waterglass | no | no (carbonate) | partly – Ca < 10 mg/L fits | protects metals but not cement; no Ca or alkalinity |
| Molde process / micronised marble | yes | yes | no – they need coagulation, which this plant does not have (assumption A5) | – |

*Adds Ca / alkalinity:* M08 slide 'Calcium AND alkalinity' – CaCO₃ dissolution (marble) and lime are the only steps that add both.

## 2.2 The choice: CO₂ dosing + marble filter

**Reasons.**
1. The network contains cast iron and cement: the water needs **calcium and alkalinity**, not only a higher pH. That rules out NaOH alone and waterglass.
2. There is no coagulation in the plant, so the Molde process and micronised marble do not fit.
3. Between CO₂ + marble and lime + CO₂, marble is simpler to operate: no lime saturator, no dust, and the filter cannot overdose – it dissolves CaCO₃ only as long as the water is undersaturated. NV 212 §12.2.2.2 names exactly this situation (little other treatment) as the field of use. In Norway about 245 plants use marble filters (M08, slide 'Three strategies in Norway').
4. The raw water has little CO₂ (C<sub>T</sub> 0.18 mmol/L), so CO₂ must be **dosed** before the filter: marble dissolves only with CO₂ (CaCO₃ + CO₂ + H₂O → Ca²⁺ + 2 HCO₃⁻).

## 2.3 Where in the plant?

> screen → **CO₂ dosing** → **marble filter** → UV disinfection → clear-water tank → network

* CO₂ is injected into the pipe just before the filter, so that it is dissolved when the water reaches the marble.
* The marble filter comes **before UV**: a new or freshly backwashed marble bed releases fine particles; UV needs clear water (NV 212 dimensions UV on the UV transmittance, §11.3). The pH itself does not affect UV (M08, slide 'Where to raise the pH?').
* If chlorine were used instead of UV, the order matters more: chlorine works much better at low pH – see B6.
* Backwash water for the marble filter comes from the clear-water tank (filtered water; normal practice according to NV 212 §4.7.4).

- CO₂ + marble filter fits a plant with little other treatment (NV 212 §12.2.2.2); soft water with little CO₂ needs CO₂ dosing first.

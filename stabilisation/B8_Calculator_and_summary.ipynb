## 8.2 Reading the summary like a reviewer

A reviewer reads a design summary column by column:

1. **Value and unit** – is the order of magnitude plausible? (e.g. CO₂ dose 10–20 mg/L for soft surface water, NV 212 Table 12.1)
2. **Source** – is it a guideline value (with table/section), a calculation, or an assumption?
3. **Assumptions** – are they reasonable, and would the result change much if they were wrong? The calculator is the tool for that question.

Try it: which result changes most when the raw-water alkalinity changes from 0.10 to 0.20 mmol/L? Which results do not change at all when the CO₂ price doubles?

## 8.3 From the example to your own design

Part C will give you a template for the water of your semester project. Until then, the checklist below shows what a design of a corrosion-control step should contain:

| Part | Content | In this example |
| --- | --- | --- |
| design basis | flows (Q<sub>avg</sub>, Q<sub>max</sub>, Q<sub>min</sub>), raw water over the year, temperature, network materials, targets with source | B1 |
| choice of method | options, criteria, decision with reasons, place in the treatment train | B2 |
| chemistry | target point, doses, treated water at min/max temperature, comparison with NV 212 | B3 |
| units | dimensioning with the critical case, guideline values with table numbers, checks | B4 (B7) |
| chemicals | consumption, dosing range, storage | B5 |
| costs | clearly marked as estimates, with price sources and what is not included | B5 |
| operation | control, monitoring, start-up, residuals, link to disinfection | B6 |
| summary | one table: value, unit, source; assumption register | B8 |

```python
sv.choice_check("In the summary table, which value would a reviewer ask you to justify <b>most</b>, because it is an assumption with a large effect?",
    ["molar mass of CaCO<sub>3</sub>", "the raw-water alkalinity range", "the freeboard of 0.3 m", "the definition of Q<sub>dim</sub>"],
    "the raw-water alkalinity range",
    "The alkalinity range sets the marble dissolution and the highest CO<sub>2</sub> dose, and so the dosing capacity – and it is an assumption until monitoring data exist.",
    {"molar mass of CaCO<sub>3</sub>": "A physical constant (NV 212 Table 4.1) – nothing to justify.",
     "the freeboard of 0.3 m": "A guideline value (NV 212 Table 4.16) with little effect on the design.",
     "the definition of Q<sub>dim</sub>": "Defined in NV 212 ch. 2 – cite it; the flows themselves are the assumption."})
```

```python
sv.show_answer('Doubling the CO₂ price: which values in the summary change, and which do not?',
    "Only the cost lines (CO₂ cost, total, NOK/m³) change. Doses, filter size, consumption and storage are technical results and independent of prices. This is why costs belong at the end of a design, and why a price estimate cannot make a design 'wrong' – it can only make it more or less expensive.")
```

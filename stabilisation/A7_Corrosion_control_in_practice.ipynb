## 7.1 Three strategies in Norway

| Strategy | Variants | Typical use |
| --- | --- | --- |
| **1 Carbonation** | CO₂ + marble filter (± NaOH) · lime + CO₂ · CO₂ stripping + marble (groundwater) | surface water with little other treatment; groundwater |
| **2 Coagulation + carbonation** | Fe coagulation + three-media filter with marble (**Molde process**) · separate marble filter · micronised marble + CO₂ + NaOH | humic surface water that needs coagulation anyway |
| **3 Waterglass** (sodium silicate) | dosing after (partial) CO₂ stripping | small plants, soft water |

About 245 Norwegian plants use marble filters and about 200 dose waterglass; carbonation dominates. Raising only the pH with NaOH is acceptable only for an all-plastic network (lecture slide "Three strategies in Norway").

## 7.2 pH adjustment: NaOH and lime

**NaOH** adds alkalinity only: CO₂ + NaOH → Na⁺ + HCO₃⁻. Cheap and simple, but it adds no calcium, and the water stays undersaturated.

**Lime + CO₂** adds both calcium and alkalinity: 2 CO₂ + Ca(OH)₂ → Ca²⁺ + 2 HCO₃⁻. Typical doses for Norwegian water: **20–35 mg/L Ca(OH)₂ and 20–40 mg/L CO₂** (NV 212/2015). Lime is delivered as quicklime (CaO) or hydrated lime (Ca(OH)₂). Quicklime is first **slaked** with water (CaO + H₂O → Ca(OH)₂), which releases heat. Because lime dissolves poorly, it is dosed as clear, saturated **lime water**, produced in a **lime saturator** – a tank in which a lime slurry is kept in contact with water long enough to dissolve, and undissolved particles settle out. Saturated lime water contains about 1.7 g Ca(OH)₂/L at 20 °C (NV 212/2015 quotes ≈ 1.5 g/L). Lime handling (slaking, saturator, clogging, dust) makes it more demanding to operate than a marble filter; it is common in coagulation plants, where the pH must be adjusted anyway.

## 7.3 CO₂ + marble filter

Marble (CaCO₃) dissolves only as long as the water is undersaturated – a **self-limiting** process that cannot overdose:

$$\mathrm{CaCO_3 + CO_2 + H_2O \rightarrow Ca^{2+} + 2\,HCO_3^-}$$

CO₂ is dosed before the filter to make the soft water aggressive enough to dissolve the marble that is needed; for CO₂-rich groundwater, part of the CO₂ is stripped first (chapter A2). Design values (NV 212/2015, §12.2.2):

| Parameter | Value |
| --- | --- |
| grain size | 1–2.5 mm, < 2 % below 0.3 mm |
| empty-bed contact time (EBCT) | 20–30 min |
| filtration rate | upflow 6–8 m/h |
| backwash | 50–60 m/h with air scour (up to 90 m/h without) |
| CO₂ dose | from Table 12.1 (raw-water alkalinity or CO₂ acidity, pH, calcium) |
| NaOH after the filter | rarely needed |

What the terms mean:

* **EBCT (empty-bed contact time)** = filter bed volume / flow. It is the time the water would need to pass the bed if the bed contained no grains; the real contact time is shorter (by the porosity).
* **Upflow / downflow:** the water enters at the bottom and flows up through the marble, or enters at the top and flows down.
* **Backwash:** the filter is cleaned by reversing the flow at a high velocity, so that the bed expands and trapped particles and fine marble grains are flushed out. **Air scour** blows air through the bed before or during backwashing to loosen the grains, so less water and a lower velocity are needed.
* **CO₂ acidity** (used in Table 12.1 for raw water with pH 5–6): the amount of base needed to raise the pH to 8.3 – in practice almost the same as the free CO₂ in mmol/L.

The filter does not reach full equilibrium in 20–30 min – the dissolution rate drops as the water approaches saturation. That is why dosing tables come from experience, and why the equilibrium calculation is an upper limit.

**Marble consumption** equals the CaCO₃ dissolved: 45 mg/L at 100 L/s is about 390 kg/d, or 140 t per year, which has to be refilled regularly.

## 7.4 Coagulation + carbonation: the Molde process

Humic surface water usually needs coagulation to remove colour (NOM): a metal salt (iron or aluminium) is added, which forms flocs that bind the humic substances and are then removed in a filter (lecture on coagulation). Coagulation with Fe³⁺ consumes alkalinity and produces CO₂:

$$\mathrm{Fe^{3+} + 3\,HCO_3^- \rightarrow Fe(OH)_3 + 3\,CO_2}$$

5 mg Fe/L consumes 0.27 mmol/L alkalinity – more than the soft raw water has – and the pH drops to about 4 (good for NOM removal with iron, which works best below pH 4.5). The CO₂ formed (about 12 mg/L) stays in the water. In the **Molde process**, a layer of marble below the anthracite and sand of the filter uses this CO₂: marble dissolves, calcium and alkalinity rise, and the residual iron is captured in the alkaline layer. One filter removes particles and stabilises the water.

* With Fe coagulation, little extra CO₂ is needed: 0–10 mg/L, typically 4–5 mg/L. With Al (coagulation pH 6.0–6.5), CO₂ doses are as for a separate marble filter.
* Three-media filter (Molde): ≤ 7.5 m/h, EBCT ≥ 20 min, marble depth ≥ 2.5 m, backwash ≥ 60 m/h. The contact time in the marble, not particle removal, limits the filtration rate.
* Drawbacks: tall filters, high backwash rates, 10–15 % loss of marble fines; a two-stage variant (separate marble filter) allows higher rates and easier backwash.
* **Micronised marble** (0.2–1 µm slurry, 25–50 mg/L) with CO₂ (10–20 mg/L) dissolves fast enough without a marble filter, but only in coagulation plants, where undissolved particles are caught; NaOH (5–15 mg/L) sets the final pH.

**What the slides show.** *Left – the Molde process:* raw water from the lake receives CO₂ and the iron coagulant (JKL = iron chloride sulphate) and flows directly onto the filter – there is no separate flocculation or sedimentation tank. In the **marble layer**, CaCO₃ dissolves and the **pH rises** (blue line); dissolved and colloidal iron that has passed the upper layers precipitates at the higher pH and is captured, so the **iron concentration falls** (red line). One filter thus removes humus and iron and stabilises the water.

## 7.5 Waterglass (sodium silicate)

Waterglass (Na₂O·nSiO₂) is a base: it raises the pH and forms a thin, self-limiting silicate layer on iron and copper corrosion products, and it complexes Fe²⁺/Mn²⁺ (less discoloured water). It adds **no calcium and no alkalinity in the carbonate sense** – it protects metals, not cement. Practice (NV 212/2015, NV 274/2022): target pH 7.7–8.5 with 5–15 mg SiO₂/L; start with a low dose and increase over months (a high start dose loosens old deposits); not for water with Ca > 10 mg/L; strip CO₂-rich water to 8–12 mg/L first. About 200 mostly small Norwegian plants use it.

## 7.6 Choosing a method

| Situation | Typical choice | Why |
| --- | --- | --- |
| surface water, little other treatment, or after **nanofiltration** (a dense membrane that removes humus but also most calcium and magnesium, so the permeate must be re-hardened) | CO₂ + marble filter | simple, self-limiting; adds Ca and alkalinity |
| humic surface water, coagulation needed | Molde process or micronised marble | one plant, both goals |
| coagulation plant that needs pH control | lime + CO₂ | pH for coagulation and stabilisation |
| groundwater with high free CO₂ | CO₂ stripping + marble filter | uses the CO₂ already there |
| small plant, Ca < 10 mg/L, iron/copper pipes | waterglass | easy dosing, less discolouration |
| all-plastic network | NaOH (pH only) | nothing to protect |

And always: **where in the train is the pH raised relative to chlorination?** Above pH 8 the C·t (chlorine concentration × contact time) needed for viruses doubles (chapter A1).

## Key points

<div style='background:rgba(67,160,71,0.12);border-left:4px solid #43a047;padding:8px 14px'>

- Norway: carbonation (CO₂ + marble, lime + CO₂, stripping + marble), coagulation + carbonation (Molde, micronised marble), waterglass.
- Marble filters are self-limiting; design by contact time (EBCT 20–30 min) and experience tables (NV 212 Table 12.1).
- Fe coagulation consumes alkalinity and produces CO₂ – the Molde process uses it to dissolve marble in the same filter.
- NaOH or waterglass add no calcium; they protect metals but not cement.
- The choice depends on raw water, other treatment, plant size and pipe materials – and must be coordinated with disinfection.

## 4.1 What alkalinity measures

**Alkalinity is the capacity of a water to neutralise strong acid.** It is measured by **titration**: in the laboratory a known volume of water (e.g. 100 mL) is stirred while a strong acid of known concentration (e.g. 0.02 mol/L HCl or H₂SO₄) is added drop by drop from a burette, and the pH is followed with an electrode. When the pH has fallen to the **end point, pH ≈ 4.5**, the added acid has used up all the water's bases. The alkalinity is the amount of acid added per litre of sample:

$$\mathrm{Alk} = \frac{c_{\mathrm{acid}} \cdot V_{\mathrm{acid}}}{V_{\mathrm{sample}}} \qquad \text{e.g. } \frac{0.02\ \mathrm{mmol/mL} \times 5.0\ \mathrm{mL}}{0.100\ \mathrm{L}} = 1.0\ \mathrm{mmol/L}$$

Why pH 4.5? At that point practically all bicarbonate and carbonate have been converted to CO₂\* – the end point you saw in chapter A3, where H⁺ and HCO₃⁻ meet. In natural water without other bases:

$$\mathrm{Alk} = [\mathrm{HCO_3^-}] + 2\,[\mathrm{CO_3^{2-}}] + [\mathrm{OH^-}] - [\mathrm{H^+}]$$

Every term counts the protons the species can take up before pH 4.5: HCO₃⁻ one, CO₃²⁻ two, OH⁻ one; free H⁺ already present counts negative. Between pH 6.5 and 8.5, HCO₃⁻ is by far the largest term, so **Alk ≈ [HCO₃⁻]** – the approximation used on the lecture slides.

**Units.** Alkalinity counts how many protons (H⁺) the water can take up, not how many molecules it contains. A bicarbonate ion takes up one proton, a carbonate ion two – so 1 mmol CO₃²⁻ counts as 2 mmol of alkalinity. Such "proton-counting" amounts are called **equivalents**: 1 mmol HCO₃⁻ = 1 meq, 1 mmol CO₃²⁻ = 2 meq (meq = milliequivalent). Norwegian practice (and Norsk Vann) writes alkalinity in "mmol/L", meaning mmol of monovalent charge per litre – the same number as **meq/L**. Other units you will meet:

| | 1 mmol/L alkalinity equals |
| --- | --- |
| as bicarbonate | 61 mg HCO₃⁻/L |
| as calcium carbonate (USA, many textbooks) | 50 mg CaCO₃/L |

Typical values: Norwegian surface water 0.05–0.2 mmol/L, target after treatment 0.6–1.0 mmol/L, hard European water 3–6 mmol/L.

(The second titration end point, pH 8.3, gives the *p-alkalinity* – carbonate and hydroxide only. In drinking water treatment it is almost always zero and rarely used.)

## 4.2 Why alkalinity and C<sub>T</sub> – and not pH?

Two waters with the same alkalinity, at pH 6.0 and pH 8.0, are mixed 1 : 1. The pH of the mixture is **not 7.0, but about 6.3** (5 °C, alkalinity 0.5 mmol/L). The reason: the pH 6.0 water contains 2.1 mmol/L C<sub>T</sub> – four times more than the pH 8.0 water (0.5 mmol/L) – because most of its carbon is free CO₂. The mixture inherits that CO₂.

pH is a logarithmic *intensity*; it does not add up. **Alkalinity and C<sub>T</sub> are amounts: they are conserved and simply add** when waters are mixed or chemicals dosed. Each treatment step changes them in a fixed way:

| Step | Alkalinity (mmol/L) | C<sub>T</sub> (mmol/L) | Ca (mmol/L) |
| --- | --- | --- | --- |
| CO₂ dosing, per mmol CO₂ | 0 | +1 | 0 |
| CO₂ stripping, per mmol CO₂ | 0 | −1 | 0 |
| NaOH, per mmol | +1 | 0 | 0 |
| lime Ca(OH)₂, per mmol | +2 | 0 | +1 |
| soda Na₂CO₃, per mmol | +2 | +1 | 0 |
| CaCO₃ (marble) dissolved, per mmol | +2 | +1 | +1 |
| strong acid, per mmol H⁺ | −1 | 0 | 0 |
| Fe³⁺ or Al³⁺ coagulant, per mmol metal | −3 | 0 | 0 |

The recipe for every calculation in this course is therefore: **convert to alkalinity and C<sub>T</sub> → add the doses → convert back to pH.** Chapter A6 turns this table into arrows.

## 4.3 Calculating the pH

From chapter A3, with Alk ≈ [HCO₃⁻] and C<sub>T</sub> ≈ [CO₂\*] + [HCO₃⁻] (valid between about pH 6 and 8.5):

$$\mathrm{pH} = \mathrm{p}K_1 + \log\frac{[\mathrm{HCO_3^-}]}{[\mathrm{CO_2^*}]} \approx \mathrm{p}K_1 + \log\frac{\mathrm{Alk}}{C_T - \mathrm{Alk}}$$

and the other way round, $C_T \approx \mathrm{Alk}\,(1 + 10^{\,\mathrm{p}K_1 - \mathrm{pH}})$. Outside this range – or when you need more accuracy – the full equation with all species is solved numerically; that is what `carbonate.py` does for every figure in these notebooks.

## 4.4 Buffer intensity and buffer capacity

Two different questions, often mixed up:

* **Buffer intensity β** – how much base (or acid) is needed to change the pH by *one unit, right here*? It is the slope of the titration curve. For the carbonate system it is **smallest around pH 8.3–8.5** (between the two p$K$s), exactly where the target pH lies. Near the target, even well-buffered water reacts strongly to small doses: with alkalinity 1.0 mmol/L at pH 8.3, only 0.05 mmol/L NaOH (2 mg/L) raises the pH by about 0.8 units. Precise dosing and good pH control matter.
* **Buffer capacity** – how much acid does it take to push the pH *down* far, e.g. from 8.3 to 7? That is what the alkalinity provides. A water with alkalinity 1.0 mmol/L needs about 0.23 mmol/L acid for that drop; a water with 0.1 mmol/L needs only 0.02 mmol/L – ten times less. In the network, acid is produced continuously: by the **biofilm** (the thin layer of bacteria growing on every pipe wall), whose respiration releases CO₂; by **nitrification**, where bacteria oxidise ammonium to nitrate and release H⁺ (NH₄⁺ + 2 O₂ → NO₃⁻ + 2 H⁺ + H₂O – rare in Norway, where chloramine is hardly used); and by corrosion reactions on materials. A low-alkalinity water then quickly loses its pH.

That is why the target asks for **alkalinity 0.6–1.0 mmol/L** in addition to pH 8.0–8.5: the pH alone says nothing about how long it will last.

## Key points

<div style='background:rgba(67,160,71,0.12);border-left:4px solid #43a047;padding:8px 14px'>

- Alkalinity = acid-neutralising capacity (titration to pH ≈ 4.5); between pH 6.5 and 8.5 Alk ≈ [HCO₃⁻]; 1 mmol/L = 61 mg HCO₃⁻/L = 50 mg CaCO₃/L.
- Alkalinity and C<sub>T</sub> are conserved and add up; pH is not. Recipe: convert → add doses → convert back.
- pH ≈ pK₁ + log(Alk / (C<sub>T</sub> − Alk)) between about pH 6 and 8.5.
- Buffer intensity is lowest around pH 8.3–8.5 – the target pH is sensitive to small doses.
- Buffer capacity (alkalinity) keeps the pH up in the network – hence the target of 0.6–1.0 mmol/L.

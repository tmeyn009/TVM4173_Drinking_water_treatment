## 3.1 Two equilibria, three species

Free CO₂ is a *diprotic* acid: it can give off two protons, one after the other.

$$\mathrm{CO_2^* \;\rightleftharpoons\; HCO_3^- + H^+} \qquad K_1 = \frac{\{\mathrm{H^+}\}\,[\mathrm{HCO_3^-}]}{[\mathrm{CO_2^*}]}$$

$$\mathrm{HCO_3^- \;\rightleftharpoons\; CO_3^{2-} + H^+} \qquad K_2 = \frac{\{\mathrm{H^+}\}\,[\mathrm{CO_3^{2-}}]}{[\mathrm{HCO_3^-}]}$$

Square brackets are concentrations in mol/L. Curly brackets denote the **activity** – the "effective" concentration that governs the equilibrium. In a solution, ions shield each other and behave as if their concentration were slightly lower; the activity is therefore a little smaller than the concentration. $\{\mathrm{H^+}\} = 10^{-\mathrm{pH}}$ is what a pH electrode measures. (Strictly, all species should be written as activities; for the soft Norwegian waters the difference is small. Chapter A5 shows when it matters.)

Equilibrium constants are often written as **pK = −log K**, just as pH = −log{H⁺}. A small pK means a strong acid.

The **total inorganic carbon** (introduced in chapter A2) is the sum of all three:

$$C_T = [\mathrm{CO_2^*}] + [\mathrm{HCO_3^-}] + [\mathrm{CO_3^{2-}}]$$

C<sub>T</sub> counts all carbon atoms that are not part of organic molecules – the carbon "budget" of the carbonate system. In laboratory reports it appears as **TIC** (total inorganic carbon) or **DIC** (dissolved inorganic carbon); it is measured by acidifying the sample and detecting the CO₂ that escapes, or calculated from pH and alkalinity (chapter A4). It must not be confused with **TOC**, the organic carbon of NOM. In the notebooks C<sub>T</sub> is given in mmol/L; in the equations above, as all concentrations, in mol/L.

These reactions are **fast** – within a fraction of a second the three species adjust to any change in pH. (The only slower step is the hydration of dissolved CO₂, which takes seconds to about a minute in cold water – still fast compared with any treatment step.) We can therefore always treat the carbonate system as being in equilibrium.

## 3.2 The constants depend on temperature

| | 0 °C | 5 °C | 10 °C | 25 °C |
| --- | --- | --- | --- | --- |
| p$K_1$ | 6.58 | **6.52** | 6.46 | 6.35 |
| p$K_2$ | 10.63 | **10.55** | 10.49 | 10.33 |

(Plummer & Busenberg 1982.) Textbooks usually give values at 25 °C. Norwegian water is 2–10 °C, so we use the values at about 5 °C – as in the lecture. The difference of 0.2 units looks small, but it shifts all calculated pH values by the same amount; for saturation (chapter A5) the error is even larger.

## 3.3 Which species dominates? The fractions α

From the two equilibria, each species can be expressed as a fraction of C<sub>T</sub> that depends only on the pH (with $h = \{\mathrm{H^+}\}$):

$$\alpha_0 = \frac{[\mathrm{CO_2^*}]}{C_T} = \frac{h^2}{h^2 + K_1 h + K_1 K_2}, \qquad
\alpha_1 = \frac{[\mathrm{HCO_3^-}]}{C_T} = \frac{K_1 h}{h^2 + K_1 h + K_1 K_2}, \qquad
\alpha_2 = \frac{[\mathrm{CO_3^{2-}}]}{C_T} = \frac{K_1 K_2}{h^2 + K_1 h + K_1 K_2}$$

<details><summary>How these are derived (click)</summary>

From $K_1$: $[\mathrm{HCO_3^-}] = [\mathrm{CO_2^*}]\,K_1/h$. From $K_2$: $[\mathrm{CO_3^{2-}}] = [\mathrm{HCO_3^-}]\,K_2/h = [\mathrm{CO_2^*}]\,K_1K_2/h^2$. Insert both in C<sub>T</sub>:
$C_T = [\mathrm{CO_2^*}]\,(1 + K_1/h + K_1K_2/h^2)$. Divide, and multiply numerator and denominator by $h^2$.
</details>

Three rules are enough to read the figure below:

* **At pH = p$K_1$** CO₂\* and HCO₃⁻ are equal (50 % each). At pH = p$K_2$ HCO₃⁻ and CO₃²⁻ are equal.
* **One pH unit** away from a p$K$ the ratio of the two neighbours is 10 : 1, two units away 100 : 1.
* **Between about 7.5 and 9.5 almost all of C<sub>T</sub> is HCO₃⁻** – the whole range of treated drinking water.

Raw Norwegian surface water (pH 6–7) lies close to p$K_1$: roughly half of its C<sub>T</sub> is free CO₂. Treated water (pH 8–8.5) contains almost only bicarbonate.

**The log-concentration diagram.** On a linear scale, small concentrations disappear – but a "small" CO₃²⁻ concentration decides whether calcium carbonate precipitates (chapter A5). On a logarithmic scale every species is visible. The lines are almost straight: each species is flat (≈ log C<sub>T</sub>) where it dominates and falls with slope 1 per pH unit beyond the neighbouring p$K$ (slope 2 beyond the next one). The lines for H⁺ and OH⁻ are added for comparison: log[H⁺] = −pH, and [OH⁻] follows from the **ion product of water**, $K_w$ = {H⁺}{OH⁻} = 10⁻¹⁴·⁷ at 5 °C (10⁻¹⁴·⁰ at 25 °C).

**Two special points** you will meet again in chapter A4, where the alkalinity is measured by titration:

* In a pure CO₂ solution, [H⁺] ≈ [HCO₃⁻] – in the log diagram, where the H⁺ line crosses the HCO₃⁻ line (pH ≈ 4–5, depending on C<sub>T</sub>).
* In a pure bicarbonate solution, [CO₂\*] ≈ [CO₃²⁻] – exactly between the two p$K$s: pH ≈ (p$K_1$ + p$K_2$)/2 ≈ 8.3 at 25 °C (8.5 at 5 °C).

These are the end points of the alkalinity titration (pH ≈ 4.5 and 8.3).

## Key points

<div style='background:rgba(67,160,71,0.12);border-left:4px solid #43a047;padding:8px 14px'>

- The carbonate system has three species: CO₂*, HCO₃⁻ and CO₃²⁻; their sum is C<sub>T</sub>.
- K<sub>1</sub> and K<sub>2</sub> fix their ratios at a given pH; at 5 °C pK<sub>1</sub> = 6.52 and pK<sub>2</sub> = 10.55.
- At pH = pK two neighbouring species are equal; each pH unit away changes their ratio 10-fold.
- Raw Norwegian water (pH 6–7): about half of C<sub>T</sub> is free CO₂. Treated water (pH 8–8.5): almost only HCO₃⁻.
- Use constants at the water temperature; the log diagram shows small but important concentrations (CO₃²⁻).

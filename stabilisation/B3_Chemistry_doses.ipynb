## 3.1 The design idea: two vectors to a target point

Chapter A6 showed that every treatment step is a vector in the Deffeyes diagram. For CO₂ + marble there are exactly two:

* **CO₂ dosing** adds carbon only: ΔC<sub>T</sub> = CO₂ dose (mmol/L), ΔAlk = 0 (arrow to the right).
* **Marble dissolution**, CaCO₃ + CO₂ + H₂O → Ca²⁺ + 2 HCO₃⁻, adds per mmol CaCO₃: 2 meq alkalinity, 1 mmol C<sub>T</sub> and 1 mmol Ca (arrow with slope 2).

Choose a target point (pH<sub>t</sub>, Alk<sub>t</sub>). Two unknowns, two equations:

$$\text{marble: } m = \frac{\mathrm{Alk_t} - \mathrm{Alk_{raw}}}{2} \qquad\qquad \text{CO}_2\text{: } d = C_{T,t} - C_{T,raw} - m$$

where C<sub>T,t</sub> is the total inorganic carbon of the target water, calculated from pH<sub>t</sub> and Alk<sub>t</sub> at the design temperature (chapter A4). All amounts in mmol/L; × 100.1 gives mg CaCO₃/L, × 44.0 gives mg CO₂/L (molar masses, NV 212 Table 4.1).

## 3.2 Choosing the target point

**Alkalinity 1.0 mmol/L.** NV 274 gives 0.6–1.0 mmol/L with ≈ 1.0 as the aim (M08, 'Target water quality'); NV 212 Table 12.1 is made for ≈ 1.0 mmol/L and ≈ 20 mg Ca/L. A higher alkalinity gives a better buffer in the network. With this choice calcium follows automatically: m = (1.00 − 0.10)/2 = 0.45 mmol/L, so Ca rises by 0.45 × 40.1 = 18.0 mg/L to **20.5 mg/L** – inside 15–25 mg/L.

**pH 8.3 – reached in the coldest water (2 °C).** The doses fix C<sub>T</sub> and alkalinity, but the pH of that water depends on temperature: it falls by about 0.1 between 2 and 12 °C (B1 and the table in 3.4). With the target met at 2 °C, the summer water has pH 8.18 – both inside 8.0–8.5. We tie the design to winter because the marble then dissolves most slowly (dissolution rates fall with temperature), so winter is the season in which the filter must work hardest.

### Hand calculation (2 °C)

| Step | Calculation | Result |
| --- | --- | --- |
| 1 raw water | pH 6.6, Alk 0.10 mmol/L → C<sub>T</sub> (A4) | C<sub>T,raw</sub> = 0.18 mmol/L |
| 2 marble | (1.00 − 0.10)/2 = 0.45 mmol/L × 100.1 | **m = 45 mg CaCO₃/L** |
| 3 target C<sub>T</sub> | pH 8.3, Alk 1.00 mmol/L, 2 °C → C<sub>T</sub> (A4) | C<sub>T,t</sub> = 1.01 mmol/L |
| 4 CO₂ | 1.01 − 0.18 − 0.45 = 0.38 mmol/L × 44.0 | **d = 16.7 mg CO₂/L** |

After the CO₂ dose, before the filter, the water has pH ≈ 5.9 and about 20 mg/L free CO₂ – this is what the marble 'sees'.

**Comparison with NV 212 Table 12.1** (M08, slide 'CO₂ + marble filter: design'). For raw water with pH 6.5 and alkalinity 0.05–0.15 mmol/L, the table gives 11–18 and 7–14 mg CO₂/L; interpolated to 0.10 mmol/L: **9–16 mg/L**. Our 16.7 mg/L is at the upper end. Two reasons: (1) the table aims at pH 8.0–8.5 – a lower target pH needs a little more CO₂, a higher one a little less, so the range covers different targets; (2) our design point is tied to 2 °C. A calculated dose that lies inside or at the edge of the experience range is plausible; a dose far outside it would mean that an input or the calculation is wrong.

```python
dv.dose_panel()
```

**What to look for in the panel.**
* Preset *Same target, met at 12 °C*: the CO₂ dose becomes 16.5 mg/L, and in winter the pH would be 8.43. Both choices are inside the target; the design point is a decision, and it must be stated.
* Preset *Snowmelt raw water*: lower alkalinity → more marble must dissolve, and the CO₂ dose changes. The dosing equipment must cover this range (3.5).
* Preset *Too ambitious: pH 8.6*: the target point lies practically on the saturation line (SI ≈ 0 at 2 °C, slightly positive at 12 °C), and the filter would have to dissolve 100 % of what can dissolve. Dissolution stops at SI = 0 and becomes very slow just before it, so a marble filter cannot reach this point. Such a target needs NaOH after the filter.

## 3.3 How close to equilibrium must the filter get?

Our target water has **SI = -0.31** at 2 °C – it is still slightly undersaturated, so marble keeps dissolving until the water leaves the filter. If the filter were infinitely long, the water would reach equilibrium (SI = 0) with pH 8.59, alkalinity 1.03 mmol/L and 21.1 mg Ca/L (the 'equilibrium limit' in the panel). Our design asks the filter to dissolve **97 %** of what could dissolve.

That sounds comfortable, but the **pH is very sensitive** to the last few mg/L: if the filter dissolves only 40 instead of 45 mg CaCO₃/L, the outlet pH is 7.69 instead of 8.30 – below the target. The reason is the buffer minimum around pH 8.3 (chapter A4): near pH 8.3 a small change in CO₂ moves the pH a lot. Two consequences for the design:

1. The **contact time** in the filter must be long enough for the dissolution to come close to the design value even in cold water – the EBCT rule of B4 (≥ 20 min) comes from experience with Norwegian filters.
2. The **outlet pH** is the most sensitive indicator of filter performance. If it drops (marble bed too low, cold water, too much CO₂), the operator sees it first in the pH. NV 212 §12.2.2.1: NaOH after the filter is normally not needed, unless the contact time is too short.

## 3.4 The treated water over the year

The doses do not change with temperature, but the pH and the saturation do. The table shows the designed water at 2–20 °C (20 °C ≈ cold-water pipes in a house):

```python
dv.temperature_table()
```

**Reading the table.** The water is slightly undersaturated (SI -0.31 at 2 °C, -0.26 at 12 °C); it moves towards saturation as it warms. In the houses (20 °C) it is close to saturation; in hot-water systems (60 °C) it becomes supersaturated and deposits some CaCO₃ – a few mg/L at most, which is acceptable (chapter A5). A slightly negative SI in the network is normal for Norwegian carbonation plants; the protective layer forms mainly from corrosion products and CaCO₃ together (NV 212 §12.2.1).

## 3.5 Dose range for the dosing equipment

The raw water varies (assumption A4). The same target point needs different CO₂ doses:

```python
dv.dose_range_table()
```

The dose range is **10.7–20.3 mg CO₂/L**. Note a result that surprises many: at the same alkalinity, the raw water with the **higher** pH needs **more** CO₂. Higher pH at the same alkalinity means less free CO₂, i.e. a smaller C<sub>T</sub> – and the target C<sub>T</sub> is the same, so more carbon has to be added. This is exactly why the design works with C<sub>T</sub> and not with the pH alone.

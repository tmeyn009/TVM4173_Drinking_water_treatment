## 5.1 From dose to consumption

A dose in mg/L times a flow in L/s gives mg/s. A useful conversion:

$$\text{kg/d} = \text{dose (mg/L)} \times Q\ (\text{L/s}) \times 86.4 / 1000 \qquad (86\,400\ \text{s/d},\ 10^6\ \text{mg/kg})$$

**CO₂.** Not all the CO₂ gas that is dosed dissolves; some escapes. We assume a **utilisation of 90 %** (<span style='background:rgba(255,167,38,0.25);padding:0 4px;border-radius:3px'>assumption</span>; it depends on the injection system and is given by the supplier).

| | Calculation | Result |
| --- | --- | --- |
| average consumption | 16.7 mg/L × 50 L/s × 86.4/1000 / 0.90 | **80 kg/d ≈ 29 t/year** |
| maximum dosing rate | 20.3 mg/L (highest dose, B3) × 288 m³/h / 1000 / 0.90 | **6.5 kg/h** |
| minimum dosing rate | 10.7 mg/L (lowest dose) × 54 m³/h / 1000 / 0.90 | **0.64 kg/h** |
| turndown ratio | 6.5 / 0.64 | about 10 : 1 |

The dosing must work over a range of about 10 : 1. A single gas flow controller often covers only about 10 : 1 accurately, so a real design might use two dosing lines (one small, one large). **Control:** the CO₂ flow follows the water flow (feed-forward), and is trimmed by the pH measured after the injection point (feedback; pH ≈ 5.9 at design conditions, B3).

**Storage.** Liquid CO₂ is delivered by truck and stored in an insulated tank at about −20 °C and 20 bar; its density is then about 1.03 t/m³ (NIST Chemistry WebBook, saturated liquid at −20 °C). For **30 days** at average consumption (<span style='background:rgba(255,167,38,0.25);padding:0 4px;border-radius:3px'>assumption</span>: covers delivery intervals and holidays): 80 kg/d × 30 d = **2.4 t ≈ 2.3 m³**. The tank and the dosing equipment are often rented from the gas supplier (an example is the tender of FREVAR KF, Fredrikstad, for CO₂ 'including rental of tank and dosing equipment').

**Marble.** The marble that dissolves must be replaced. The product contains at least 98 % CaCO₃ (Visnes data sheet), and some fines are lost in backwash; we add 5 % (<span style='background:rgba(255,167,38,0.25);padding:0 4px;border-radius:3px'>assumption</span>; for three-media filters M08, slide 'Molde process', reports 10–15 % loss by washing out fines – a stand-alone marble filter is washed more gently):

| | Calculation | Result |
| --- | --- | --- |
| consumption | 45 mg/L × 50 L/s × 86.4/1000 / 0.98 × 1.05 | **208 kg/d ≈ 76 t/year** |
| bed loss | 208 kg/d / 1500 kg/m³ / 48 m² × 30 d | about 87 mm per month |
| refill interval | 0.3 m reserve / bed loss | about every 104 days (≈ 22 t, e.g. in 1-t big bags) |

Refill before the bed drops below the design depth of 2.7 m – otherwise the EBCT falls and so does the outlet pH (B3). Measure the bed depth monthly.

## 5.2 Running costs – ESTIMATES

Unit prices for water-treatment chemicals are not published; they are negotiated in tenders. The prices below are **estimates** for teaching and must be replaced by supplier offers in a real design:

| Item | Price used | Where the estimate comes from |
| --- | --- | --- |
| liquid CO₂, delivered, incl. tank rental | **3 NOK/kg** (range 2–5) <span style='background:rgba(229,57,53,0.22);padding:0 4px;border-radius:3px'>ESTIMATE</span> | IMARC Group reports a wholesale price for liquid CO₂ in Germany of about 219 USD/t in Q2 2026 (≈ 2.3 NOK/kg at ≈ 10.5 NOK/USD). Small deliveries, transport and tank rental come on top. |
| filter marble, delivered in big bags | **2 NOK/kg** (range 1.5–3) <span style='background:rgba(229,57,53,0.22);padding:0 4px;border-radius:3px'>ESTIMATE</span> | no public unit price found. IVAR IKS (Stavanger) awarded a filter-marble contract of 9 million NOK (TED notice 263316-2020), without stating quantities. Ask the supplier. |

With these prices:

| | Calculation | Result <span style='background:rgba(229,57,53,0.22);padding:0 4px;border-radius:3px'>ESTIMATE</span> |
| --- | --- | --- |
| CO₂ | 80 kg/d × 365 × 3 NOK/kg | 88 kNOK/year |
| marble | 208 kg/d × 365 × 2 NOK/kg | 152 kNOK/year |
| total chemicals | | **240 kNOK/year** |
| per m³ water | 240 kNOK / 1.58 million m³ | **0.15 NOK/m³** |
| first marble fill (once) | 216 t × 2 NOK/kg | 432 kNOK |

**Not included:** construction (building, filter boxes, pipes, valves), the CO₂ installation, electricity (backwash pump and blower), labour, maintenance and laboratory analyses. These are usually much larger than the chemical costs over the lifetime of a plant, but cannot be estimated responsibly here – they need cost data from a consultant or a contractor.

**What to look for.** Marble is the larger running cost although its price per kg is lower: almost three times as much marble as CO₂ is used per m³ (45 vs. 17 mg/L). The CO₂ utilisation matters: with 70 % instead of 90 % the CO₂ cost rises by almost 30 %. The cost per m³ is small compared with typical water tariffs – corrosion control is cheap compared with the damage corrosion does to pipes.

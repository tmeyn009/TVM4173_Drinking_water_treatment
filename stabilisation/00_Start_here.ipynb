### How to use the notebooks
1. Open a chapter and choose *Run → Run All Cells*. You do **not** need to write code.
2. **Interactive panels:** click a *preset* to load an example, then move the sliders. Hover over points and lines to read values; drag a rectangle to zoom, double-click to reset; the camera icon saves a figure as PNG.
3. **Check yourself:** try first, then press *Check* or *Show answer*.
4. Pictures from the lecture are shown from the course's shared folder. If one is missing, the text tells you which lecture slide to look at.
5. **Dark theme?** Boxes and text adapt automatically. For dark figures, set `sv.DARK = True` in the first code cell of a chapter and run all cells again.
6. Something looks wrong? *Kernel → Restart Kernel and Run All Cells*.

**Reading:** Norsk Vann 212/2015, ch. 12 (corrosion control) and 9.3 (hardness); Norsk Vann 274/2022 (Korrosjonsbeskyttelse – erfaring og ny kunnskap); lecture M08.

The chemistry behind all figures is in `carbonate.py` (you may open it). The example waters are typical Norwegian waters, not your semester-project waters.

### Symbols and abbreviations

Units: in the text, tables and figures concentrations are given in **mmol/L** or **mg/L**. In equations with equilibrium constants ($K$) they must be in **mol/L** (1 mmol/L = 10⁻³ mol/L).

| Symbol | Meaning | Unit | Introduced in |
| --- | --- | --- | --- |
| CO₂\* (free CO₂) | dissolved CO₂ plus the small amount of true carbonic acid H₂CO₃ | mg/L, mmol/L | A2 |
| HCO₃⁻, CO₃²⁻ | bicarbonate (hydrogen carbonate), carbonate | mmol/L | A2, A3 |
| C<sub>T</sub> | **total inorganic carbon** = CO₂\* + HCO₃⁻ + CO₃²⁻; in lab reports also TIC or DIC | mmol/L | A2, A3 |
| Alk | **alkalinity** – the ability of the water to neutralise acid, mostly bicarbonate | mmol/L (= meq/L) | A1, A4 |
| [X] | concentration of species X | mol/L in equations | A2 |
| {X} | activity ("effective concentration") of X; {H⁺} is what a pH electrode measures | mol/L | A3, A5 |
| pH | −log{H⁺} | – | |
| K<sub>H</sub>, p<sub>CO₂</sub> | Henry constant; partial pressure of CO₂ in a gas | mol/(L·atm); atm | A2 |
| K₁, K₂ | first and second acidity constant of carbonic acid | mol/L | A3 |
| K<sub>w</sub> | ion product of water, {H⁺}{OH⁻} | (mol/L)² | A3 |
| K<sub>s</sub> | solubility product of calcite, {Ca²⁺}{CO₃²⁻} at equilibrium | (mol/L)² | A5 |
| pK | −log K (e.g. pK₁ = 6.52 at 5 °C) | – | A3 |
| α₀, α₁, α₂ | fractions of C<sub>T</sub> present as CO₂\*, HCO₃⁻, CO₃²⁻ | – | A3 |
| β | buffer intensity: base needed to change the pH by one unit | mmol/L per pH | A4 |
| SI | saturation index for calcium carbonate (< 0 dissolves, > 0 can deposit) | – | A1, A5 |
| pHₛ, LSI | saturation pH; Langelier saturation index LSI = pH − pHₛ | – | A5 |
| S | ionic-strength correction in the pHₛ formula | – | A5 |
| CCPP | calcium carbonate precipitation potential: CaCO₃ that can precipitate (+) or still dissolve (−) | mg CaCO₃/L | A5 |
| I, γ, z | ionic strength; activity coefficient; charge of an ion | mol/L; –; – | A5 |
| EC | electrical conductivity | µS/cm | A5 |
| NOM, TOC | natural organic matter (humus); total organic carbon | mg C/L | A1, A3 |
| C·t | disinfection: chlorine concentration × contact time | mg·min/L | A1 |

| NV 212, NV 274 | Norsk Vann reports 212/2015 (design guideline) and 274/2022 (corrosion protection) | | A1 |

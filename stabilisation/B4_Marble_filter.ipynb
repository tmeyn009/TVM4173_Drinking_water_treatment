## 4.1 Design values from NV 212

The marble filter is a **downflow** filter (water from top to bottom), standing alone after the CO₂ dosing. NV 212 gives design values in two places; they partly overlap:

| Design value | NV 212 | Value used | Why |
| --- | --- | --- | --- |
| grain size | §12.2.2.4: 1–2.5 mm (or 1–3 / 0.5–2.0 mm), < 2 % below 0.3 mm | 1–2.5 mm | standard product |
| empty bed contact time (EBCT) | §12.2.2.4: 20–30 min; Table 12.3: ≥ 20 min | **≥ 20 min with one filter out** | cold water; the minimum even at the worst moment |
| filter rate, downflow | §12.2.2.4: typical 8–10 m/h ¹ | **8 m/h with one filter out** | lower end: cold water, longer contact |
| marble depth | Table 12.3: ≥ 2.5 m (three-media) / ≥ 2.0 m (upflow) | ≥ 2.5 m | follows from rate × EBCT |
| number of filters | Table 4.16: at least 2, preferably 4; one stand-by per 10 filters | **4** | see 4.3 |
| area per filter | Table 4.16: < 100 m² | – | check |
| freeboard | Table 4.16: about 30 cm | 0.3 m | |
| backwash (water) | Table 12.3: ≥ 60 m/h; Table 4.19: 50–65 m/h, 8–12 min (water only); §12.2.2.4: enough to expand the bed by 10 %, typically 50–60 m/h with air scour, up to 90 m/h without | **60 m/h, 10 min** | see 4.5 |

¹ In the PDF text of NV 212 the downflow rate reads "8 – 0 m/h"; the digit is lost in the layout – most likely 8–10 m/h. Table 12.3 gives ≤ 7.5 m/h for three-media (Molde) filters and ≤ 6 m/h for upflow filters; in those the marble layer is only part of the bed. Where guidelines are unclear, choose the conservative side and say so.

**Contact time and rate are linked through the depth:** EBCT = bed volume / flow = (A · H) / (A · v) = H / v. So for a given EBCT, a higher rate needs a deeper bed: H = v · EBCT.

## 4.2 The critical situation: Q<sub>max</sub> with one filter out of service

Filters are backwashed one at a time, and sometimes one is out for repair. During that time the other filters take the whole flow (NV 212 §4.7.4: consider the peak load when one filter is taken out for washing; Table 4.16: stand-by). We therefore require **EBCT ≥ 20 min and v ≤ 8 m/h at Q<sub>dim</sub> = 288 m³/h with one filter out**. In normal operation the filters are then comfortably within the rules.

### Step by step (4 filters)

| Step | Calculation | Result |
| --- | --- | --- |
| 1 bed volume in service | V = Q · EBCT = 288 m³/h × 20/60 h | 96 m³ |
| 2 area in service | A<sub>on</sub> = Q / v = 288 / 8 | 36 m² |
| 3 depth | H = v · EBCT = 8 m/h × 20/60 h = 2.67 m → round up | **H = 2.7 m** (≥ 2.5 m ✓) |
| 4 area per filter | 36 m² / 3 filters in service | **12.0 m²** (≈ 3.5 × 3.5 m) |
| 5 total area | 4 × 12.0 m² | 48 m² |
| 6 normal operation (Q<sub>max</sub>, all 4 in) | v = 288/48; EBCT = 2.7/v | 6.0 m/h; 27 min ✓ (20–30) |
| 7 average day (Q<sub>avg</sub> = 180 m³/h) | EBCT = 2.7 × 48 / 180 | 43 min |
| 8 marble | 48 m² × (2.7 + 0.3 reserve) m × 1.5 t/m³ | **216 t** first fill |

*Bulk density 1.5 t/m³:* the supplier's data sheet (Visnes Kalk, 1–3 mm water-treatment marble) gives 'density 1.45–1.55 kg/dm³'. This must be the **bulk** density of the granular bed – solid calcite has 2.71 kg/dm³ – i.e. the bed has about 45 % pores.

**Why 4 filters?** With 2 filters, the one that stays in service must take the full flow: it needs the whole 36 m², so the plant has 2 × 36 = 72 m² – 50 % more than with 4 filters (48 m²). More filters mean less spare area, but more valves, pipes and control. NV 212 Table 4.16 recommends 'at least 2, preferably 4'.

## 4.3 Heights

| Layer (from bottom) | Height | Source |
| --- | --- | --- |
| support gravel with nozzle floor | 0.3 m | <span style='background:rgba(255,167,38,0.25);padding:0 4px;border-radius:3px'>assumption</span> (NV 212 §4.7.3.8: adapted to the filter material, no number) |
| marble, design depth | 2.7 m | 4.2 |
| marble reserve, dissolves between refills | 0.3 m | <span style='background:rgba(255,167,38,0.25);padding:0 4px;border-radius:3px'>assumption</span> (see B5) |
| water above the bed | 1.5 m | <span style='background:rgba(255,167,38,0.25);padding:0 4px;border-radius:3px'>assumption</span>; NV 212 §4.7.4: downflow filters are usually designed for 2–3 m available head loss |
| freeboard | 0.3 m | NV 212 Table 4.16 |
| **filter box** | **5.1 m** | |

During backwash the bed expands by about 10 % (NV 212 §12.2.2.4), i.e. about 0.3 m – this fits into the water zone above the bed, so no marble is washed over the backwash channel. Coarse marble (1–2.5 mm) and no coagulant mean a low head loss; the water depth above the bed is ample.

## 4.4 Backwash

Marble filters accumulate fines (abraded marble, particles from the raw water) and must be backwashed: **60 m/h** water for **10 min**, preferably with air scour first (NV 212 Table 12.3: ≥ 60 m/h; Table 4.19: 50–65 m/h for 8–12 min with water only). Too low a rate lets fines accumulate and the head loss rises; too high a rate washes out marble (§12.2.2.1).

| | Calculation | Result |
| --- | --- | --- |
| backwash flow per filter | 60 m/h × 12.0 m² | **720 m³/h = 200 L/s** – the backwash pump size |
| volume per wash | 720 m³/h × 10/60 h | 120 m³ |
| per week | 4 filters × 1 wash/week (<span style='background:rgba(255,167,38,0.25);padding:0 4px;border-radius:3px'>assumption</span>) × 120 m³ | 480 m³ |
| share of production | 480 / 30 240 m³ (Q<sub>avg</sub> × 1 week) | 1.6 % |

The backwash pump (≈ 200 L/s) is more than twice Q<sub>max</sub>: it takes its water from the clear-water tank, which must hold at least one wash (here 120 m³) above the minimum level for the network. How often the filters must be washed depends on the raw water turbidity and is found in operation – start with once per week and adjust by the head loss.

## Key points

<div style='background:rgba(67,160,71,0.12);border-left:4px solid #43a047;padding:8px 14px'>

- Size the marble filter for Q<sub>dim</sub> with one filter out: here EBCT ≥ 20 min and v ≤ 8 m/h.
- EBCT = H/v: depth = rate × contact time; with 8 m/h and 20 min the bed is 2.7 m deep.
- Example: 4 filters × 12.0 m², 2.7 m marble + 0.3 m reserve, 216 t marble; normal operation 6.0 m/h and 27 min.
- Backwash 60 m/h for 10 min: 720 m³/h per filter, about 1.6 % of the production.

## 7.1 The water

A small town uses a well in a sand and gravel aquifer. The water (example water from Part A; same flows as Example waterworks for easy comparison):

| pH | Alkalinity | Calcium | Temperature | free CO₂ | C<sub>T</sub> |
| --- | --- | --- | --- | --- | --- |
| 6.4 | 0.70 mmol/L | 12 mg/L | 6 °C (constant) | 37 mg/L | 1.55 mmol/L |

This is typical for Norwegian groundwater: low pH, high free CO₂ (often 20–60 mg/L), 5–25 mg Ca/L (M08, slide 'Typical Norwegian drinking water').

## 7.2 Why strip – and why not completely?

The water already has far more CO₂ than needed. A marble filter alone would dissolve CaCO₃ until equilibrium and end at an alkalinity of about 2.3 mmol/L and 43 mg Ca/L (try the preset *Groundwater: marble only* in chapter A7) – much more than needed, and a large marble consumption. Stripping removes CO₂ without chemicals (M08, slide 'Physical treatment: CO₂ stripping': used above about 20 mg/L).

But the marble needs CO₂ to dissolve. **NV 212 §12.2.1** therefore recommends: strip to **12–15 mg/L CO₂, then marble filtration** (also M08, slides 'Physical treatment: CO₂ stripping' and 'Choose a concept').

**What the vector design says.** Using the method of B3 with the target pH 8.3 and Alk 1.0 mmol/L gives: strip 30.5 mg/L (leaving 7.0 mg/L), then dissolve 15 mg CaCO₃/L. That reaches the target exactly on paper – but with only 7.0 mg/L CO₂ left, the driving force for dissolution is small, and the last mg/L dissolve slowly (B3, 3.3): the filter would need a much longer contact time. The NV rule leaves more CO₂, so the marble dissolves quickly – at the price of a higher alkalinity than 1.0 mmol/L. That is the trade-off to discuss in a design report.

**Design choice:** follow NV 212 and strip to **13 mg/L**. Expected result (equilibrium limit of the filter): pH 8.39, alkalinity 1.28 mmol/L, Ca 23.7 mg/L – alkalinity above the NV 274 range of 0.6–1.0 mmol/L. Iron and cement benefit from more alkalinity and calcium; copper prefers less inorganic carbon (M08, slide 'Targets per pipe material'). Accepting 1.2–1.4 mmol/L is the usual practice for groundwater with this route; for a network with much copper one might strip further and accept a longer EBCT.

## 7.3 The stripping tower

In a tower aerator the water trickles over packing (25–50 mm plastic or steel bodies) while a fan blows air upwards. NV 212 gives:

| Design value | NV 212 | Value used |
| --- | --- | --- |
| efficiency E | §8.2.3.4: 90–95 % in towers | 90 % (lower end, cold water) |
| surface loading v<sub>f</sub> | §8.2.4.4: 40–100 m³/(m²·h) | 60 m³/(m²·h) |
| packing depth | §8.2.4.4: 3–5 m | 4 m |
| air-to-water ratio L/V | Table 8.2: 5–100 for towers | 10 |

**Partial-flow stripping.** E = (c<sub>in</sub> − c<sub>out</sub>)/(c<sub>in</sub> − c<sub>sat</sub>) (NV 212 §8.2.4), with c<sub>sat</sub> = 1.1 mg/L, the CO₂ in equilibrium with air at 6 °C (Henry's law, chapter A2). With E = 90 % the tower outlet has 4.8 mg/L – far below 13 mg/L. So only part of the flow passes the tower, the rest bypasses it, and the two are mixed. The share f through the tower follows from a mass balance on CO₂ (approximately; the notebook mixes C<sub>T</sub> exactly):

$$f \approx \frac{c_{in} - c_{target}}{c_{in} - c_{out}} = \frac{37 - 13}{37 - 4.8} \approx 75\ \%$$

| | Calculation | Result |
| --- | --- | --- |
| water through the towers | 75 % × 288 m³/h | 215 m³/h |
| area per tower (2 towers) | 215 / 60 / 2 | 1.8 m² → diameter 1.5 m |
| air flow | 10 × 215 m³/h | 2 154 m³/h |

**Is L/V = 10 enough?** The least air that could carry the CO₂ away follows from Henry's law: L/V<sub>min</sub> = (c<sub>in</sub> − c<sub>out</sub>)/(H · c<sub>in</sub>), with the dimensionless Henry constant H = 1/(K<sub>H</sub> R T) = 0.70 at 6 °C. Here L/V<sub>min</sub> = 1.2, so L/V = 10 gives a margin of about 8. NV 212 asks for L/V = 20–40 only for gases that are hard to strip (e.g. trichloroethene); CO₂ is easier.

**Scaling in the tower?** The tower outlet has pH ≈ 7.3 and SI ≈ -1.6 – far from supersaturation, so no CaCO₃ deposits on the packing (M08, slide 'Physical treatment: CO₂ stripping': scaling if stripping is driven too far). After blending: pH 6.86, 13 mg/L CO₂.

**Marble filter.** The rules of B4 do not depend on the dose, so the same filter (4 × 12.0 m², 2.7 m) fits. About 29 mg CaCO₃/L dissolves – marble consumption ≈ 135 kg/d at Q<sub>avg</sub>.

## Key points

<div style='background:rgba(67,160,71,0.12);border-left:4px solid #43a047;padding:8px 14px'>

- CO₂-rich groundwater: strip to 12–15 mg/L (NV 212), then the remaining CO₂ dissolves marble.
- A tower removes 90–95 %, so only part of the flow is stripped (here 75 %) and blended with the bypass.
- Size the tower with surface loading, depth and L/V from NV 212; check L/V against the Henry's-law minimum.
- The NV route gives alkalinity above 1.0 mmol/L – a trade-off between fast dissolution and the target range that must be discussed.

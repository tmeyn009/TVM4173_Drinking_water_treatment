## 2.1 Where does the CO₂ come from?

Carbon dioxide reaches the water from three sources:

* **The atmosphere** contains about 0.042 % CO₂ (420 ppm). Water in contact with air takes up CO₂ until equilibrium is reached.
* **Soil air.** In the soil, microorganisms break down organic matter and roots respire. Soil air therefore contains much more CO₂ than the atmosphere – typically 1–5 %, sometimes 10 %. Water that infiltrates through the soil and becomes **groundwater** equilibrates with this soil air and has no contact with the atmosphere afterwards.
* **Decomposition in the water itself**, for example in the deep water of a **stratified** lake. In summer (warm surface layer) and in winter (ice cover) many lakes are layered: the surface water does not mix with the cold deep water. In the dark deep layer organic matter is broken down but no photosynthesis consumes CO₂, so CO₂ accumulates – and many waterworks take their raw water from exactly this depth.

## 2.2 Henry's law

In equilibrium, the concentration of dissolved CO₂ is proportional to its partial pressure in the gas:

$$[\mathrm{CO_2^*}] = K_H \cdot p_{\mathrm{CO_2}}$$

Square brackets [ ] denote a **concentration**, here in mol/L. So $[\mathrm{CO_2^*}]$ is the concentration of dissolved CO₂ (mol/L), $p_{\mathrm{CO_2}}$ the partial pressure of CO₂ in the gas (atm; 0.042 % → $4.2 \cdot 10^{-4}$ atm), and $K_H$ the Henry constant (mol L⁻¹ atm⁻¹). $K_H$ depends strongly on temperature – **cold water holds more gas**:

| | 0 °C | 5 °C | 25 °C |
| --- | --- | --- | --- |
| $K_H$ (mol L⁻¹ atm⁻¹) | 0.078 | 0.064 | 0.034 |
| CO₂ in equilibrium with air (mg/L) | 1.4 | 1.2 | 0.6 |
| CO₂ in equilibrium with 3 % soil air (mg/L) | 103 | 85 | 45 |

To convert: 1 mmol/L CO₂ = 44.01 mg/L.

**CO₂\* – "free CO₂".** Dissolved CO₂ reacts with water to carbonic acid, CO₂ + H₂O ⇌ H₂CO₃, but only about 0.2 % is present as true H₂CO₃. Because the two cannot be distinguished in practice, they are lumped together as CO₂\* (often written H₂CO₃\*). In water treatment this is called **free CO₂**.

## 2.3 CO₂ makes water acidic

CO₂\* is a weak acid. It releases a proton (H⁺) and forms **bicarbonate** (hydrogen carbonate), HCO₃⁻ (chapter A3):

$$\mathrm{CO_2^* \rightleftharpoons HCO_3^- + H^+}$$

**Rain water** contains practically nothing but water and CO₂ from the air. In equilibrium with air its pH is about **5.6** (5.5 at 5 °C) – rain is naturally slightly acidic even without pollution.

**A groundwater** in equilibrium with soil air contains much more CO₂ and therefore has a low pH – but how low depends also on its **alkalinity**, the bicarbonate it has picked up by weathering minerals (chapter A4). The more alkalinity, the higher the pH for the same CO₂.

**Total inorganic carbon, C<sub>T</sub>.** Once dissolved, CO₂ is present in three forms: free CO₂ (CO₂\*), bicarbonate (HCO₃⁻) and – at high pH – carbonate (CO₃²⁻). The pH decides how the carbon is divided between them (chapter A3), but the **total amount of inorganic carbon** stays the same unless CO₂ is added or removed. This total is called C<sub>T</sub>:

C<sub>T</sub> = [CO₂\*] + [HCO₃⁻] + [CO₃²⁻], usually given in mmol/L.

"Inorganic" distinguishes it from the organic carbon in NOM (measured as TOC). C<sub>T</sub> is the "carbon budget" of the water: stripping CO₂ lowers it, dosing CO₂ or dissolving marble raises it.

## 2.4 CO₂ stripping (aeration)

When a CO₂-rich groundwater is brought into contact with air – in a cascade, a packed tower, by spraying or by blowing air through it – CO₂ escapes until the water approaches equilibrium with air. This is **CO₂ stripping** (degassing):

$$\mathrm{CO_2(aq) \rightarrow CO_2(g)}$$

What changes, and what does not?

* **Free CO₂ and C<sub>T</sub> decrease** – carbon leaves the water.
* **The alkalinity does not change.** CO₂ carries no charge; removing it removes no base. (Chapter A4 explains this in detail.)
* **The pH rises**, because the ratio bicarbonate / free CO₂ increases. Try the presets "Groundwater in soil air" and "…the same, stripped" above: the same water goes from pH ≈ 6.1 to ≈ 7.9.

**The limit.** Stripping can at most bring the water to equilibrium with air: about 1.2 mg/L CO₂ at 5 °C (about 0.7 mg/L at 20 °C, the value often quoted). In practice, plants stop well above this, because the remaining CO₂ is needed:

* Groundwater before a **marble filter**: strip to **12–15 mg CO₂/L** – the remaining CO₂ dissolves the marble and adds calcium and alkalinity (NV 212/2015).
* Before **waterglass** dosing: strip to 8–12 mg CO₂/L.
* Stripped too far, a water with much calcium can become supersaturated and deposit CaCO₃ in the plant.

Stripping uses no chemicals, only energy for pumping and blowing air. It is used when the raw water contains a lot of CO₂ (more than about 20 mg/L, 0.5 mmol/L).

**What the picture shows.** A simple aerator: water is sprayed into the air as a fountain. Breaking the water into droplets creates a large surface between water and air. The water contains far more CO₂ than is in equilibrium with air (e.g. 40 mg/L instead of about 1 mg/L), so CO₂ diffuses out of every droplet. The larger the contact surface and the longer the contact time, the closer the water gets to equilibrium with air. Packed towers and cascades work on the same principle, with better control of air and water flows. Note that aeration also adds oxygen – useful for iron and manganese removal in groundwater.

**Dosing CO₂** is the reverse: CO₂ gas is dissolved into the water. It increases free CO₂ and C<sub>T</sub> and lowers the pH, again without changing the alkalinity. That sounds like the wrong direction – but CO₂ is needed to dissolve marble (CaCO₃ + CO₂ + H₂O → Ca²⁺ + 2 HCO₃⁻). Chapters A6 and A7 show why it pays off.

## Key points

<div style='background:rgba(67,160,71,0.12);border-left:4px solid #43a047;padding:8px 14px'>

- Free CO₂ (CO₂*) comes from the air, from soil air (1–5 % CO₂) and from decomposition in the water.
- Henry's law: [CO₂*] = K<sub>H</sub>·p<sub>CO₂</sub>; cold water holds more CO₂ (≈ 1.2 mg/L from air at 5 °C).
- CO₂ is an acid: rain in air has pH ≈ 5.6; groundwater in soil air has a low pH unless it has much alkalinity.
- Stripping removes CO₂: C<sub>T</sub> down, pH up, **alkalinity unchanged**; the limit is equilibrium with air.
- Before a marble filter, groundwater is stripped only to 12–15 mg/L – the rest is used to dissolve marble.

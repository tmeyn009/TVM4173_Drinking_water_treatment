## 5.1 Solubility of calcium carbonate

Calcium carbonate (calcite – the mineral in marble and limestone) dissolves until the product of the calcium and carbonate **activities** reaches the **solubility product** $K_s$. (Activity = "effective concentration", slightly lower than the concentration because ions shield each other; written with curly brackets { }. Section 5.3 shows how it is calculated.)

$$\mathrm{CaCO_3(s) \rightleftharpoons Ca^{2+} + CO_3^{2-}} \qquad K_s = \{\mathrm{Ca^{2+}}\}\,\{\mathrm{CO_3^{2-}}\}$$

| | 5 °C | 25 °C | 60 °C |
| --- | --- | --- | --- |
| p$K_s$ | 8.39 | 8.48 | 8.76 |

Unusually, calcite becomes **less soluble when the water is warmer** – the reason for scale in kettles, hot-water heaters and heat exchangers.

Because the CO₃²⁻ concentration rises tenfold per pH unit (chapter A3), the saturation depends strongly on pH: at the same calcium and alkalinity, raising the pH by one unit raises the ion product about tenfold.

## 5.2 Saturation index SI and Langelier index LSI

The **ion product** (IAP, ion activity product) is {Ca²⁺}{CO₃²⁻} of the actual water – it can be smaller or larger than $K_s$. Only in equilibrium is it equal to $K_s$. The **saturation index** compares the two:

$$\mathrm{SI} = \log\frac{\{\mathrm{Ca^{2+}}\}\{\mathrm{CO_3^{2-}}\}}{K_s} \qquad \begin{cases} \mathrm{SI} < 0 & \text{undersaturated: dissolves CaCO}_3\text{ (aggressive)} \\ \mathrm{SI} = 0 & \text{in equilibrium} \\ \mathrm{SI} > 0 & \text{supersaturated: can deposit CaCO}_3 \end{cases}$$

**Langelier (1936)** expressed the same idea as a pH difference. The **saturation pH**, pHₛ, is the pH at which the water – with its calcium and bicarbonate – would be exactly saturated. Combining $K_2$ and $K_s$:

$$\mathrm{pH_s} = \mathrm{p}K_2 - \mathrm{p}K_s - \log[\mathrm{Ca^{2+}}] - \log[\mathrm{HCO_3^-}] + S \qquad \mathrm{LSI} = \mathrm{pH} - \mathrm{pH_s}$$

Concentrations in mol/L, [HCO₃⁻] ≈ alkalinity. **S** is a small correction for the ionic strength of the water, explained in 5.3 (set S = 0 if you ignore it). For drinking water (pH 6.5–8.5) LSI and SI are practically equal.

<details><summary>Where the pHₛ formula comes from (click)</summary>

At saturation $\{\mathrm{Ca^{2+}}\}\{\mathrm{CO_3^{2-}}\} = K_s$. From $K_2$: $\{\mathrm{CO_3^{2-}}\} = K_2\{\mathrm{HCO_3^-}\}/\{\mathrm{H^+}\}$. Insert and solve for $\{\mathrm{H^+}\}$: $\{\mathrm{H^+}\}_s = K_2\{\mathrm{Ca^{2+}}\}\{\mathrm{HCO_3^-}\}/K_s$. Take −log, write activities as γ·concentration; the γ terms form $S$.
</details>

## 5.3 Ionic strength: the correction $S$

Ions in solution shield each other; their effective concentration (activity) is lower than the analytical one: $\{i\} = \gamma_i [i]$. The activity coefficients follow from the **ionic strength** $I = \tfrac12 \sum c_i z_i^2$, which can be estimated from the conductivity ($I \approx 1.6\cdot10^{-5}\times$ EC in µS/cm). With the Davies equation:

$$\log\gamma = -A\,z^2\left(\frac{\sqrt I}{1+\sqrt I} - 0.3\,I\right), \qquad S = -\log(\gamma_{\mathrm{Ca}}\,\gamma_{\mathrm{HCO_3}})$$

The symbols: $c_i$ is the concentration of ion *i* (mol/L) and $z_i$ its charge (2 for Ca²⁺, 1 for HCO₃⁻ and Na⁺); the sum runs over all ions in the water. **EC** is the electrical conductivity, measured with a conductivity meter in µS/cm. $A$ is a constant that depends slightly on temperature (0.49 at 5 °C, 0.51 at 25 °C). Because $z$ enters squared, the effect for divalent Ca²⁺ is four times larger than for monovalent ions. In treated water (200 µS/cm, $I$ ≈ 3 mmol/L), **$S$ ≈ 0.13** – the same size as the difference between "just saturated" and "slightly aggressive". In very soft raw water it is below 0.05.

## 5.4 CCPP: how much, not only which direction

SI and LSI tell the *direction*. The **calcium carbonate precipitation potential (CCPP)** tells the *amount*: how many mg CaCO₃/L the water would precipitate (CCPP > 0) or dissolve (CCPP < 0) before it reaches equilibrium. It has to be calculated iteratively, because precipitation changes calcium, alkalinity, C<sub>T</sub> and pH at the same time.

Two waters with the same SI can differ a lot: at SI = −0.5, a soft water (Alk 0.2 mmol/L, 5 mg Ca/L) can dissolve only about 3 mg/L CaCO₃, a harder water (Alk 2 mmol/L, 40 mg Ca/L) about 11 mg/L.

## 5.5 Calcium – the hidden third variable

The Deffeyes diagram (chapter A6) has two axes, alkalinity and C<sub>T</sub>. Saturation, however, also depends on **calcium**. A saturation line in the diagram is therefore valid for **one calcium concentration only**. When marble dissolves, calcium rises together with alkalinity and C<sub>T</sub> – the water moves, and so does its saturation line. Equilibrium is reached where they meet. Reading the end point of a marble filter from a fixed saturation line gives the wrong answer; the calculation must follow calcium along the way.

## 5.6 What saturation does – and does not – tell you

* A slightly positive SI helps to form and maintain a thin protective layer of calcium carbonate and corrosion products on iron and cement surfaces.
* SI says nothing about **how fast** CaCO₃ dissolves or precipitates.
* SI says nothing about **copper and lead**. Their release depends on pH, carbonate (copper forms soluble carbonate complexes), NOM and stagnation time (NV 274/2022).
* Use constants at the **water temperature**: p$K_2$ − p$K_s$ is 2.16 at 5 °C but 1.85 at 25 °C – an error of 0.3 pH units in pHₛ.

## Key points

<div style='background:rgba(67,160,71,0.12);border-left:4px solid #43a047;padding:8px 14px'>

- Calcite dissolves until {Ca²⁺}{CO₃²⁻} = K<sub>s</sub>; it is less soluble in warm water (scaling).
- SI = log(ion product / K<sub>s</sub>); LSI = pH − pHₛ with pHₛ = pK₂ − pKₛ − log[Ca] − log[HCO₃⁻] + S.
- The ionic-strength correction S (≈ 0.1 in treated water) and the temperature of the constants matter.
- CCPP gives the amount (mg CaCO₃/L) that can dissolve or precipitate – SI only the direction.
- Calcium is the hidden third variable: a saturation line belongs to one Ca concentration. SI says nothing about copper or lead.

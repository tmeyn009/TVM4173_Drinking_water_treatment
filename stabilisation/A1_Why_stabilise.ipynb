## 1.1 Norwegian raw water: soft, low-buffered and often CO₂-rich

About 90 % of the Norwegian population gets its drinking water from **surface water** (lakes and rivers), 10 % from groundwater (FHI). Most of Norway's bedrock is granite, gneiss and other hard, slowly weathering rock with little limestone. Rain and snowmelt therefore pick up very few dissolved minerals on their way to the lake. A typical Norwegian surface water is

* **very soft:** 1–5 mg Ca/L (hard European waters: 50–100 mg/L and more),
* **low in alkalinity:** often below 0.2 mmol/L – there is almost no buffer,
* **slightly acidic:** pH 6–7,
* **rich in natural organic matter** (humus, NOM): brown colour and organic acids.

<div style='border-left:4px solid #1f77b4;padding:4px 12px;margin:8px 0'>

**Three words you will meet in every chapter**

* **Hardness** is the content of calcium and magnesium. *Soft* water has little of both. Norwegian surface water with 1–5 mg Ca/L contains only 0.02–0.12 mmol/L calcium.
* **Alkalinity** is the ability of a water to neutralise acid. In natural water it consists almost entirely of **bicarbonate**, HCO₃⁻, dissolved from rocks and soil. It is given in mmol/L. Chapter A4 explains it in detail.
* **Buffer.** A water with alkalinity resists changes in pH: added acid is used up by the bicarbonate before the pH falls much. A water with little alkalinity has little buffer – a small amount of acid or base changes its pH a lot.
</div>

**Groundwater** is less common, but important for many smaller municipalities. It has passed through soil and sediments, where it picks up CO₂ from the decomposition of organic matter (chapter A2). Typical: calcium 5–25 mg/L, often iron and manganese, a low pH and a high content of **free CO₂, often 20–60 mg/L**.

Both types of water have the same problem: they are **corrosive** to most pipe materials – they dissolve and attack them.

## 1.2 What corrosion does

**Iron and steel** (cast iron, ductile iron, steel, galvanised pipes) corrode electrochemically. At one place on the pipe wall iron dissolves (anode), at another place oxygen is reduced (cathode):

$$\text{anode: } \mathrm{Fe \rightarrow Fe^{2+} + 2\,e^-} \qquad \text{cathode: } \mathrm{O_2 + 2\,H_2O + 4\,e^- \rightarrow 4\,OH^-}$$

The dissolved iron is oxidised and precipitates as iron hydroxides and oxides – rust. In soft, low-alkalinity water these deposits are loose and porous; they grow into **tubercles** that narrow the pipe and are torn loose when the flow changes: **red water** at the tap. In water with enough calcium, carbonate and buffer, a denser layer of corrosion products and calcium carbonate forms, which slows further corrosion.

**Copper** dissolves faster at low pH and high free CO₂. Copper at the tap gives blue-green stains in sinks and a metallic taste; above 2 mg/L it is a health concern. **Lead** comes mainly from older brass fittings and solder; it dissolves in the same kind of water and is toxic even at µg/L levels.

**Cement-based materials** (concrete pipes, cement mortar linings in ductile iron pipes, asbestos cement) contain calcium hydroxide and calcium carbonate. Aggressive water dissolves them: the material weakens, and in very soft water the pH in the network can rise sharply after contact with fresh linings.

**Plastic pipes** (PE, PVC, PP) are not attacked by the water chemistry. But almost every network also contains metals – in fittings, valves, house installations – so the water must still be made non-corrosive.

**What the pictures show.** *Left:* the inside of a corroded cast-iron pipe. The brown-orange bumps are **tubercles** – mounds of iron oxides and hydroxides that grow over corroding spots. They narrow the pipe, increase the pressure loss and break off when the flow changes: that is the "red water" customers see. This is the typical damage caused by soft, aggressive Norwegian water. *Middle and right:* the **opposite** problem in hard water – a thick, white layer of calcium carbonate (scale) in a pipe and on a tap. It forms when water contains more calcium carbonate than it can hold. Corrosion control aims between the two: just enough calcium carbonate for a thin protective layer, but no thick scale.

**Consequences of corrosion** (lecture slide "Problems due to corrosion"):

| for the network | for the water | for the operator |
| --- | --- | --- |
| shorter pipe life, leaks, water loss | red water, turbidity, iron | complaints |
| narrower pipes, higher pressure loss | copper and lead at the tap | flushing and cleaning |
| contamination through leaks | taste, staining | high maintenance costs |

In Norway scaling is rare, but it is the reason why the target is "just saturated", not "as much calcium carbonate as possible" – see 1.4 and chapter A5.

## 1.3 What the rules require

The Norwegian drinking water regulation (*drikkevannsforskriften*) does not prescribe a treatment, but sets requirements at the tap:

| Parameter | Requirement | Comment |
| --- | --- | --- |
| pH | 6.5–9.5 (action limit) | Mattilsynet no longer recommends a pH target |
| corrosivity | "the water shall not be corrosive" | judged from pH, chloride, sulphate, conductivity |
| copper | 2 mg/L | ≤ 1 mg/L recommended (staining) |
| lead | 10 µg/L | the revised EU Drinking Water Directive lowers this to **5 µg/L by 2036**; Norway is preparing to adopt it |

The regulation says *what* (not corrosive), but not *how much* calcium or alkalinity. The numbers come from the water sector: **Norsk Vann report 274/2022**.

## 1.4 The target water quality

To protect the network, the water leaving the treatment plant should have (Norsk Vann 274/2022):

| Parameter | Target | Why |
| --- | --- | --- |
| pH | **8.0–8.5** | above the saturation pH of CaCO₃; low copper release |
| alkalinity | **0.6–1.0 mmol/L** (≈ 1.0) | buffer – keeps the pH stable in the network |
| calcium | **15–25 mg Ca/L** (≈ 20) | together with carbonate: a thin CaCO₃ layer, saturation index ≈ 0 |
| *or* silicate | pH 7.7–8.5 and 10–15 mg SiO₂/L | waterglass (sodium silicate, a base that forms a thin protective film on metal surfaces; chapter A7) instead of calcium – small plants |

**Saturation in a few words.** Calcium carbonate (CaCO₃, the mineral in limestone and marble) dissolves in water until the water is *saturated* – it then holds exactly as much as it can, and the water is in equilibrium with the solid. A water that can still dissolve CaCO₃ is *undersaturated* (aggressive: it attacks cement and cannot build a protective layer); a water that holds more than it can keep is *supersaturated* and deposits scale. The **saturation pH** is the pH at which a water, with its calcium and alkalinity, would be exactly saturated. The **saturation index SI** tells how far a water is from this point: SI < 0 aggressive, SI = 0 in equilibrium, SI > 0 deposits. Chapter A5 shows how both are calculated.

Compare the target with the raw water in 1.1: the pH must go up, but above all **calcium and alkalinity must be added** – often ten times what the raw water contains. Stabilisation in Norway is therefore mostly *the opposite of softening*. (An older recommendation, pH 8–9, is still found in NV 212/2015; it has been replaced by 8.0–8.5.)

## 1.5 One water, many materials – a compromise

Different pipe materials prefer different water (lecture slide "Targets per pipe material"):

| Material | prefers | because |
| --- | --- | --- |
| copper | high pH (≈ 8.5), **low** carbonate, free CO₂ < 1–2 mg/L | carbonate forms soluble copper complexes |
| cast iron | pH 7.5–8.5, **high** alkalinity and calcium | a dense layer of carbonate and corrosion products |
| cement | pH 8.0–8.5, alkalinity ≈ 1 mmol/L, Ca ≈ 20 mg/L | aggressive water leaches calcium |

A network contains all of them, so the target in 1.4 is a **compromise**. In addition, flow conditions (stagnant water in houses at night), NOM and pipe age all influence the metal release (NV 274/2022).

## 1.6 Stabilisation and disinfection belong together

Free chlorine is present as hypochlorous acid HOCl (a strong disinfectant) and hypochlorite OCl⁻ (much weaker). Which form dominates depends on the pH: at pH ≈ 7.5 – the p$K_a$ of HOCl, the pH at which half is HOCl and half OCl⁻ (the "p$K$" idea is explained in chapter A3) – both are equal; above pH 8 most of the chlorine is OCl⁻.

Chlorine disinfection is designed with the **C·t value**: the free-chlorine concentration C (mg/L) multiplied by the contact time t (min). The C·t needed for **3-log inactivation** (99.9 % of the viruses inactivated) at 4 °C **doubles** when the pH is raised from below 7 to above 8: from 4 to 8 mg·min/L (Norsk Vann 202/2014). Either chlorinate *before* the pH is raised, or design the contact tank for the larger C·t. UV disinfection is not affected by pH.

## Key points

<div style='background:rgba(67,160,71,0.12);border-left:4px solid #43a047;padding:8px 14px'>

- Norwegian raw water is soft (1–5 mg Ca/L), low in alkalinity (< 0.2 mmol/L) and slightly acidic; groundwater is often rich in free CO₂.
- Such water corrodes iron (red water), copper and lead-containing fittings, and leaches cement.
- Regulation: pH 6.5–9.5 and "not corrosive"; Cu 2 mg/L, Pb 10 µg/L (5 µg/L in the revised EU directive).
- Target (NV 274/2022): pH 8.0–8.5, alkalinity 0.6–1.0 mmol/L, Ca 15–25 mg/L – calcium and alkalinity must be **added**.
- The target is a compromise between materials, and raising the pH doubles the chlorine C·t for viruses.

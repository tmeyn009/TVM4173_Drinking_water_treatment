## 6.1 Control

* **CO₂:** the dosing rate follows the water flow (feed-forward) and is corrected by the pH after the injection point (feedback; set point ≈ 5.9, B3). The set point is adjusted seasonally or by the outlet pH.
* **Marble filter:** has no control loop – it dissolves CaCO₃ as long as the water is undersaturated. What the operator controls is the bed depth (refill) and the backwash.
* **Outlet pH** is the key indicator (B3, 3.3): it reacts first when the bed becomes too shallow, the water very cold, or the CO₂ dose too high.

## 6.2 Monitoring plan

Frequencies are <span style='background:rgba(255,167,38,0.25);padding:0 4px;border-radius:3px'>assumption</span>s typical for a plant of this size; the legal requirements for sampling follow from the drinking water regulations and Mattilsynet's guidance.

| What | Where | How often | Why |
| --- | --- | --- | --- |
| flow | inlet | online | dosing control, consumption |
| pH | after CO₂ injection | online | CO₂ dosing control |
| pH | after marble filter | online, with alarm < 8.0 | filter performance (B3) |
| turbidity | after marble filter | online | fines after backwash; UV needs clear water |
| head loss | each filter | online | backwash timing |

| bed depth | each filter | monthly | refill before the depth is below 2.7 m |
| CO₂ tank level | tank | online | ordering |
| copper, iron at the tap | network, sampling points | as required by the regulations | effect of the corrosion control |

## 6.3 Start-up

New marble contains fines (the specification allows < 2 % below 0.3 mm, NV 212 §12.2.2.4). Before the plant goes into operation the filters are backwashed until the water is clear, and the first filtrate after each backwash is run to waste until the turbidity is low (NV 212 §4.7.4: 'modningsvann' to drain). The outlet pH reaches its design value only after some hours of operation; check alkalinity and calcium in the first weeks.

## 6.4 Backwash water

Each wash produces about 120 m³ of water with marble fines and the particles from the raw water. It is collected in a backwash water tank, where the particles settle. The clear water can be returned to the inlet or discharged; the sludge goes to the sewer or is collected. Which route is allowed depends on the local discharge permit (<span style='background:rgba(255,167,38,0.25);padding:0 4px;border-radius:3px'>assumption</span>: return of the clear water to the inlet).

## 6.5 The link to disinfection

**UV.** The pH does not affect UV disinfection (M08, slide 'Where to raise the pH?'), but two side effects of the marble filter do: *fines* after backwash lower the UV transmittance, on which the UV plant is designed (NV 212 §11.3) – hence turbidity monitoring and filtering to waste; and *CaCO₃ deposits* on the quartz sleeves of the lamps if the water is supersaturated (SI > 0). Our water has SI -0.31 to -0.26 – little risk; the UV reactor's wipers handle the rest. The marble filter without coagulant is not counted as a hygienic barrier in this design (<span style='background:rgba(255,167,38,0.25);padding:0 4px;border-radius:3px'>assumption</span>): UV alone is the barrier against microorganisms.

**Chlorine.** If chlorine is used after the marble filter, the pH matters a lot. Free chlorine is present as HOCl (a strong disinfectant) and OCl⁻ (much weaker): HOCl ⇌ OCl⁻ + H⁺. The pK<sub>a</sub> is about 7.5 at 25 °C and 7.80 at 2 °C (Morris 1966). At pH 8.3 and 2 °C only 24 % of the free chlorine is HOCl, against 94 % at the raw-water pH 6.6. NV 212 Table 11.1 (and M08, slide 'pH and disinfection') therefore requires a **twice as high C·t** for 3-log virus inactivation above pH 8 as below pH 7 (8 instead of 4 mg·min/L at 4 °C).

**Two options** (M08, slide 'Where to raise the pH?'): (1) chlorinate **before** the pH is raised, e.g. before the marble filter – the contact time in the filter then counts as part of the chlorine contact time, but the filter must be checked for that use; or (2) chlorinate **last** and design the contact tank for the larger C·t. Corrosion control and disinfection must be designed together.

## Key points

<div style='background:rgba(67,160,71,0.12);border-left:4px solid #43a047;padding:8px 14px'>

- CO₂ is dosed flow-proportionally with pH trim; the marble filter needs no control, but bed depth and backwash must be managed.
- The outlet pH is the key indicator of filter performance; turbidity protects the UV step.
- With chlorine, raising the pH doubles the C·t needed above pH 8 – chlorinate before the pH increase or enlarge the contact tank.
- Backwash water and its route must be planned and permitted.

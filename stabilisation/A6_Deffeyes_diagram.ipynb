## 6.1 The diagram

In the previous chapters a water was described by many numbers: pH, free CO₂, bicarbonate, carbonate, alkalinity, total inorganic carbon. Two of them are enough to describe the whole carbonate system – all others can be calculated:

* **C<sub>T</sub>, the total inorganic carbon** (x-axis): *how much carbonate "material" the water contains* – all carbon atoms present as CO₂\*, HCO₃⁻ or CO₃²⁻, in mmol/L. It changes only when carbon enters or leaves the water: CO₂ dosing or stripping, dissolving or precipitating CaCO₃.
* **Alkalinity** (y-axis): *how much acid the water can neutralise* – essentially how much of that carbon is present as bicarbonate and carbonate rather than free CO₂, in mmol/L (= meq/L). It changes only when acid or base is added.

Kenneth Deffeyes (1965) proposed plotting these two against each other. Because both are conserved and add up when chemicals are dosed (chapter A4), **every water is a point** in the diagram, and **every treatment step moves the point along a straight line** – a **vector**: an arrow whose direction is fixed by the chemical and whose length is proportional to the dose.

**How to read the diagram.** The pH is not an axis – it is read from the grey **pH lines**, each of which connects all waters with the same pH. A point between two lines has a pH between those values. Example: the soft surface water (C<sub>T</sub> = 0.18 mmol/L, alkalinity 0.10 mmol/L) lies between the lines for pH 6.5 and 7.0, closer to 6.5 – its pH is 6.6. Moving to the right (more C<sub>T</sub> at the same alkalinity, i.e. more free CO₂) crosses lines of lower pH; moving up (more alkalinity at the same C<sub>T</sub>) crosses lines of higher pH. The green **target zone** contains all waters with pH 8.0–8.5 and alkalinity 0.6–1.0 mmol/L.

**Where the pH lines come from.** Write the alkalinity (chapter A4) with the species fractions α₁ (share of C<sub>T</sub> present as HCO₃⁻) and α₂ (share present as CO₃²⁻) from chapter A3; [OH⁻] and [H⁺] are the small contributions of water itself:

$$\mathrm{Alk} = C_T\,(\alpha_1 + 2\alpha_2) + [\mathrm{OH^-}] - [\mathrm{H^+}]$$

At a fixed pH the α's are constant, so alkalinity is (almost) proportional to C<sub>T</sub>: the pH lines are straight lines through (nearly) the origin. Low pH → flat lines (much C<sub>T</sub> for little alkalinity, i.e. much free CO₂); around pH 8.3 the slope is ≈ 1 (Alk ≈ C<sub>T</sub>, almost all carbon is bicarbonate). The lines lie close together between pH 8 and 9 – the weak buffer intensity of chapter A4, seen geometrically. The diagram depends on temperature, because $K_1$ and $K_2$ do.

## 6.2 Treatment steps as vectors

From the table in chapter A4:

| Step | Direction | Why |
| --- | --- | --- |
| NaOH | ↑ straight up | alkalinity +1, C<sub>T</sub> unchanged |
| lime Ca(OH)₂ | ↑ straight up (twice as long per mmol) | alkalinity +2, plus calcium – invisible in the diagram |
| CO₂ dosing | → right | C<sub>T</sub> +1, alkalinity unchanged |
| CO₂ stripping | ← left | C<sub>T</sub> −1 |
| strong acid, Fe/Al coagulant | ↓ straight down | alkalinity −1 (acid) or −3 per mmol metal |
| CaCO₃ dissolution (marble) | ↗ slope 2 | alkalinity +2, C<sub>T</sub> +1 |
| soda Na₂CO₃ | ↗ slope 2 | like CaCO₃, but without calcium |

## 6.3 A graphical design: CO₂ + marble

**Task:** bring the soft surface water (pH 6.6, alkalinity 0.10 mmol/L, C<sub>T</sub> 0.18 mmol/L, 5 °C) to pH 8.3 and alkalinity 1.0 mmol/L with CO₂ and a marble filter.

1. **Find the target point.** At 5 °C, pH 8.3 and Alk = 1.0 mmol/L correspond to C<sub>T</sub> ≈ 1.01 mmol/L.
2. **The marble vector** must supply the missing alkalinity: 1.0 − 0.10 = 0.90 mmol/L → 0.45 mmol/L CaCO₃ (45 mg/L), which also adds 0.45 mmol/L C<sub>T</sub> and 0.45 mmol/L Ca (18 mg/L).
3. **The CO₂ vector** supplies the rest of the C<sub>T</sub>: 1.01 − 0.18 − 0.45 = 0.38 mmol/L CO₂ ≈ **17 mg/L**.
4. **Check saturation.** The end point has 20.5 mg Ca/L, but SI ≈ −0.3: pH 8.3 is still below pHₛ. The water would dissolve a little more marble – in practice the filter is designed for contact time, and the final pH ends slightly higher or a little NaOH is dosed.

**Graphically, in the panel below** (preset "Soft water → pH 8.3, Alk 1.0"):

1. Find the raw water: the black point marked *raw*, at C<sub>T</sub> 0.18 and alkalinity 0.10.
2. Find the target: where the pH 8.3 line crosses the height alkalinity = 1.0 – the blue square.
3. From the target, go **backwards** along a line with slope 2 (two units down for one unit left) – the direction of marble dissolution – until you reach the horizontal line through the raw water (alkalinity 0.10).
4. That intersection is where **arrow 1** (CO₂, horizontal to the right) must end. Its length on the x-axis is the CO₂ dose in mmol/L (× 44 = mg/L). **Arrow 2** (marble) runs from there to the target; its height divided by 2 is the CaCO₃ dose in mmol/L (× 100 = mg/L).
5. Check the dashed saturation line for the final calcium: if the target lies below it, the water is still undersaturated (SI < 0).

The panel does this construction for any target you choose.

## Key points

<div style='background:rgba(67,160,71,0.12);border-left:4px solid #43a047;padding:8px 14px'>

- Deffeyes diagram: alkalinity versus C<sub>T</sub>; pH lines are nearly straight lines through the origin and depend on temperature.
- Each treatment step is a vector: base ↑, acid/coagulant ↓, CO₂ dosing →, stripping ←, CaCO₃ ↗ (slope 2).
- Design = add vectors from the raw water to the target point; the doses are the vector lengths.
- Calcium is invisible in the diagram – check saturation (SI) separately at the end point.
